# Predicting House Prices

*The goal of this project is to develop a Linear Regression model that predicts the price of a house (in $1000s) based purely on numerical property and neighborhood attributes.

*The dataset represents a collection of residential properties, each described by measurable quantitative factors such as size, number of rooms, age, and proximity to key facilities. The challenge is to model the relationship between these features and the house price, capturing how each factor contributes to property valuation.

*Objective:
Build and evaluate a regression model that:

●Accurately predicts the house price.
●Identifies the most influential numerical factors affecting the price.
●Minimizes prediction error using evaluation metrics like R², MAE, and RMSE.

```python
#Checking Skewness of X variables
X = df.drop("price", axis=1)
X.hist(bins=10)
plt.show()
for i in X.columns:
    print(f"{i} : {skew(X[i])}")
```

```python
from scipy.stats import skew
#finding the skewness in each variable
data_num_skew = X.apply(lambda i: skew(i.dropna()))
#subsetting the variables that are highly skewed
data_num_skewed = data_num_skew[(data_num_skew > .75) | (data_num_skew < -.75)]
data_num_skew
data_num_skewed
```
